**Ficha de datos para el TP6**

- **Pregunta:** ¿Cómo se distribuye la morosidad en el sistema financiero argentino a agosto 2026?

- **Fuente (URL):** Central de Deudores del Sistema Financiero, BCRA.

- **Método de obtención:** descarga. Cada mes se baja un zip con un archivo txt de ancho fijo (deudores.txt, unos 7 GB) con su LEAME_DEUDORES.pdf. Los archivos de ARCA (Actividades_ARCA.txt y Padron_ARCA.txt) también se descargan. El procesamiento lo hago con Python + DuckDB y guardo el resultado en Parquet (procesar_deudores.py y cruzar_deudores.py). Ambos los obtengo mediante una declaración jurada personal en ARCA.

- **Unidad de observación y período:** un registro por deudor–entidad–periodo. Agosto 2026.

- **Condiciones de uso o licencia:** datos públicos elaborados por el BCRA.

- **Cómo voy a verificar los datos:**
  - Controlo que el conteo de registros coincida con el archivo original y que todas las líneas tengan 171 caracteres.
  - Reviso que el cruce con las entidades cubra el 100%.
  - Comparo totales por entidad y por situación con las estadísticas agregadas que publica el BCRA.
  - Reviso nulos, duplicados y consistencia entre períodos consecutivos.
