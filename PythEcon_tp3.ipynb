````python
%pip install pandas

import pandas as pd

# Leemos la base completa
potencia = pd.read_csv(
    "potencia_instalada_mod.csv",
    encoding="latin1",
    sep="|",
    index_col="Unnamed: 0"
)

# Quitamos el nombre del índice
potencia.index.name = None

# Definimos las columnas
columnas = ["periodo", "central", "agente", "agente_descripcion", "region", "categoria_region", "fuente_generacion", "tecnologia", "potencia_instalada_mw"]

# Guardamos una copia de las columnas en df
df = potencia[columnas].copy()

# Mostramos la cantidad de filas y columnas de df
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

# Verificamos que el dataframe tiene 15.793 filas y 9 columnas.

# Chequeamos si las columnas tienen valores faltantes
df.info()
# Las columnas agente y agente_descripción presentan 4 valores faltantes cada una. El resto de las columnas no.
````

**(2)** Agreguen a `df` una columna `fecha` de tipo `datetime`, creada a partir de `periodo`, y una columna `periodo_formateado` con el formato `2020-10` (año-mes, como texto). Comprueben que la primera fecha de la base sea octubre de 2015. Si el resultado es el esperado, eliminen de `df` la columna `periodo` original.

````python
# Convertimos a fecha
df["fecha"] = pd.to_datetime(
    df["periodo"],
    format="%d/%m/%Y %H:%M"
)

# Creamos una columna de texto con el formato año-mes
df["periodo_formateado"] = df["fecha"].dt.strftime("%Y-%m")

# Control: chequeamos formato de primera fecha
primera_fecha = df["fecha"].iloc[0]
print("Primera fecha:", primera_fecha)

# Si la primera fecha es la esperada, eliminamos periodo. Sino, verificar código
if primera_fecha == pd.Timestamp("2015-10-01"):
    df = df.drop("periodo", axis=1)
    print("Fecha correcta")
else:
    print("Fecha incorrecta. Verificar")

# Control: vemos las dos columnas nuevas
df[["fecha", "periodo_formateado"]].head()
````

````python
### (a) ###

%pip install matplotlib

import matplotlib.pyplot as plt

# Potencia instalada total para cada fecha
total_antes = (
    df.groupby(by="fecha")
    .agg({"potencia_instalada_mw": "sum"})
    .reset_index()
)

# Ordenamos fechas de menor a mayor
total_antes = total_antes.sort_values(by="fecha")

# Graficamos evolución del total del país
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    total_antes["fecha"],
    total_antes["potencia_instalada_mw"],
    marker="o",
    markersize=3
)

ax.set_title("Potencia instalada total antes de la limpieza")
ax.set_xlabel("Período")
ax.set_ylabel("Potencia instalada (MW)")
ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()

# Cinco períodos con mayor potencia total
print(
    total_antes.sort_values(
        by="potencia_instalada_mw",
        ascending=False
    ).head()
)
## Los períodos en los que se detectan los outliers son: 01-01-2020, 01-09-2019 y 01-07-2019
````

````python
### (b) ###

# Comparamos la potencia por central entre un mes normal (06-2019) y uno conflictivo (07-2019), y la que muestre mayor variación de potencia entre un mes y otro podría ser candidata para la comparación posterior

# Potencia total por central en los dos meses a comparar
junio = (
    df[df["periodo_formateado"] == "2019-06"]
    .groupby("central")["potencia_instalada_mw"].sum()
)

julio = (
    df[df["periodo_formateado"] == "2019-07"]
    .groupby("central")["potencia_instalada_mw"].sum()
)

# Calculamos las diferencias
comparacion = pd.DataFrame()
comparacion["potencia_junio"] = junio
comparacion["potencia_julio"] = julio
comparacion["aumento_mw"] = comparacion["potencia_julio"] - comparacion["potencia_junio"]

# Ordenamos de mayor a menor el aumento entre un mes y otro
comparacion = comparacion.sort_values(
    by="aumento_mw",
    ascending=False
)

# Mostramos las 10 primeras
display(comparacion.head(10))

# Como la central YACYHI fue la de mayor aumento entre un mes y otro, miramos sus filas para ver si hay componentes nuevos o si hay observaciones duplicadas

# Guardamos filas de YACYHI correspondientes al mes normal (06-2019)
junio_yacyhi = df[
    (df["central"] == "YACYHI")
    & (df["periodo_formateado"] == "2019-06")
]

# Guardamos filas de YACYHI correspondientes al mes sospechoso (07-2019)
julio_yacyhi = df[
    (df["central"] == "YACYHI")
    & (df["periodo_formateado"] == "2019-07")
]

# Mostramos todas las columnas de ambos meses
print("JUNIO")
print(junio_yacyhi)

print("JULIO")
print(julio_yacyhi)
````

Como vimos al principio, que se repita la central y el mes no alcanza para decir que hay un registro repetido. Es por ello que tomamos una central y mes particular, y comparamos el resto de los datos, y en particular, tecnología y potencia instalada. Así, verificamos que en julio de 2019 hay un registro repetido para la central YACYHI.

````python
### (c) ###

# En (b) encontramos un registro repetido para la central YACYHI. Esto podría estar ocurriendo para el resto de las centrales en las fechas que vimos eran conflictivas.

# Identificamos las filas duplicadas
eliminadas = df[df.duplicated()]

# Filas a eliminar por mes y central
resumen_eliminadas = (
    eliminadas.groupby(["periodo_formateado", "central"])
    .size()
    .reset_index(name="filas_eliminadas")
)

print("Filas a eliminar por mes y central:")
print(resumen_eliminadas)

# Aquí vemos que las filas duplicadas provienen de las fechas conflictivas, y para todas las centrales

# Total de filas duplicadas por mes
print("\nTotal de filas a eliminar por mes:")
print(eliminadas.groupby("periodo_formateado").size())

# Eliminamos las filas duplicadas y conservamos la original en df
df_limpio = df.drop_duplicates().copy()

# Chequeamos cuántas filas quedaron
print("\nFilas originales:", len(df))
print("Filas después de limpiar:", len(df_limpio))
print("Filas eliminadas:", len(df) - len(df_limpio))
````

En el inciso (c) encontramos que el problema del registro repetido visto en una central en (b), se repite para todas las centrales en las 3 fechas que habíamos identificado como conflictivas, por lo que se eliminan estos duplicados, quedándonos con una base limpia.

````python
### (d) ###

# Graficamos evolución de la potencial total con la base limpia
total_despues = (
    df_limpio.groupby(by="fecha")
    .agg({"potencia_instalada_mw": "sum"})
    .reset_index()
)

# Ordenamos por fecha
total_despues = total_despues.sort_values(by="fecha")

# Graficamos los totales corregidos
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    total_despues["fecha"],
    total_despues["potencia_instalada_mw"],
    marker="o",
    markersize=3
)

ax.set_title("Potencia instalada total después de la limpieza")
ax.set_xlabel("Período")
ax.set_ylabel("Potencia instalada (MW)")
ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
````

````python
### (e) ###

# Queremos encontrar los meses que faltan en la serie. Para ello, creamos primero un calendario con todos los meses entre el inicio y el final
meses = pd.date_range(
    start=df_limpio["fecha"].min(),
    end=df_limpio["fecha"].max(),
    freq="MS"
)

# Lo expresamos como string año-mes
meses = meses.strftime("%Y-%m")

# Obtenemos los meses que aparecen en la base, sin repetirlos
meses_presentes = df_limpio["periodo_formateado"].unique()

# Buscamos los meses faltantes
meses_faltantes = []

for mes in meses:
    if mes not in meses_presentes:
        meses_faltantes.append(mes)

print("Cantidad de meses faltantes:", len(meses_faltantes))
print("Meses faltantes:", meses_faltantes)
````

**(6)** Construyan `total_periodo`: un DataFrame con una fila por período, ordenado por período, con las columnas `periodo_formateado`, `potencia_instalada_mw` (el total del país) y `variacion_pct` (la variación porcentual respecto de la fila anterior).

Con ese DataFrame, armen un gráfico con **dos ejes verticales**: barras con `variacion_pct` (eje izquierdo) y una línea con el nivel de potencia instalada (eje derecho).

_Nota: el método `.shift()` aplicado a una columna les puede servir como paso intermedio. Tengan en cuenta lo que encontraron en el ejercicio 3(e): ¿todas las variaciones que calcularon son mensuales? Respondan en una celda de texto._

````python
# Sumamos la potencia instalada del país para cada período
total_periodo = (
    df_limpio.groupby("periodo_formateado")
    .agg({"potencia_instalada_mw": "sum"})
    .reset_index()
)

# Ordenamos cronológicamente
total_periodo = total_periodo.sort_values(by="periodo_formateado")

# Calculamos la variación porcentual respecto de la fila anterior
total_periodo["variacion_pct"] = (total_periodo["potencia_instalada_mw"]/ total_periodo["potencia_instalada_mw"].shift(1)- 1) * 100

# Creamos la figura y el eje izquierdo
fig, ax1 = plt.subplots(figsize=(12, 5))

# En el eje izquierdo se muestra con un gráfico de barras las variaciones porcentuales de la potencia instalada
ax1.bar(
    total_periodo["periodo_formateado"],
    total_periodo["variacion_pct"],
    color="steelblue",
    label="Variación (%)"
)

ax1.set_xlabel("Período")
ax1.set_ylabel("Variación respecto del período anterior (%)", color="black")
ax1.tick_params(axis="x", rotation=90, labelsize=8)
ax1.axhline(0, color="gray", linewidth=0.8)

# Creamos el eje derecho
ax2 = ax1.twinx()

# En el eje derecho se muestra con un gráfico de línea la evolución en el nivel de potencia instalada
ax2.plot(
    total_periodo["periodo_formateado"],
    total_periodo["potencia_instalada_mw"],
    color="darkred",
    label="Potencia instalada (MW)"
)

ax2.set_ylabel("Potencia instalada (MW)", color="black")
ax2.set_ylim(0, 45000)

# Título y leyendas
ax1.set_title("Potencia instalada total y variación porcentual")
ax1.legend(loc="upper left")
ax2.legend(loc="upper right")

plt.tight_layout()
plt.show()
````

En el ejercicio 3(e) vimos que hay meses faltantes en la base, por lo que las variaciones porcentuales no son necesariamente mensuales sino que el cálculo compara cada observación con la anterior disponible. Es por ello que, cuando hay un mes faltante, la variación corresponde a un intervalo de dos meses o más.
